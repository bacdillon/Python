# Load the book

```python
# Open the file "Tan Tock Seng article.txt" in read mode ("r"), using UTF-8 text encoding
# The "with" block automatically closes the file once we're done reading it
with open("Tan Tock Seng article.txt", "r", encoding="utf-8") as file:
    book = file.read()  # Read the entire file content into one big string and store it in "book"
```

```python
type(book)  # Check what kind of object "book" is (confirms it's a string, i.e. <class 'str'>)
```

# How many chapters?

### With string methods

```python
book.count("Chapter")  # Count how many times the literal word "Chapter" appears anywhere in the text
```

### With regex

```python
import re  # Import Python's built-in regular expressions module, needed for pattern matching
```

```python
# Build a pattern that matches the word "Chapter" followed by a space and one or more digits (e.g. "Chapter 12")
pattern = re.compile("Chapter [0-9]+")
re.findall(pattern, book)  # Find every match of that pattern in the text and return them as a list
```

```python
# Same idea, but [a-z]+ matches lowercase letters instead of digits after "Chapter " (likely a mistake/test)
pattern = re.compile("Chapter [a-z]+")
re.findall(pattern, book)  # This will find nothing useful, since chapters are numbered, not lettered
```

```python
# Repeat the correct digit-based pattern from before
pattern = re.compile("Chapter [0-9]+")
findings = re.findall(pattern, book)  # Store all matches (e.g. "Chapter 1", "Chapter 2", ...) in "findings"
len(findings)  # Count how many chapter headings were found -> gives the total number of chapters
```

# Which are the sentences where "Singapore" & "Healthcare" was used?

```python
# Build a pattern to capture a full sentence that mentions "Singapore":
#   [A-Z]{1}        -> sentence must start with exactly one capital letter
#   [^.]*           -> any characters that are not a period (rest of the sentence so far)
#   [^a-zA-Z]+      -> at least one non-letter character (e.g. a space) right before "Singapore"
#   Singapore       -> the literal word "Singapore" (capital S only, case-sensitive)
#   [^a-zA-Z]+      -> at least one non-letter character right after "Singapore"
#   [^.]*.          -> the rest of the sentence up to and including the closing period
pattern = re.compile("[A-Z]{1}[^.]*[^a-zA-Z]+Singapore[^a-zA-Z]+[^.]*.")
findings = re.findall(pattern, book)  # Get every sentence in the book matching that pattern
findings  # Display the list of matched sentences
```

```python
# Same kind of sentence-matching pattern, but:
#   - starts with [^.]* instead of requiring a capital first letter (looser at the start)
#   - matches "singapore" in lowercase in the pattern text
#   - re.IGNORECASE makes the match case-insensitive, so it catches "Singapore", "SINGAPORE", "singapore", etc.
pattern = re.compile("[^.]*[^a-zA-Z]*singapore[^a-zA-Z]+[^.]*.", re.IGNORECASE)
findings = re.findall(pattern, book)  # Find all matching sentences, regardless of capitalization
len(findings)  # Count how many sentences mention "Singapore" in any capitalization
```

```python
# Same sentence-capturing pattern style as before, but looking for "Healthcare" (case-sensitive, capital H)
pattern = re. compile("[A-Z]{1}[^.]*[^a-zA-Z]+Healthcare[^a-zA-Z]+[^.]*.")
findings = re.findall(pattern, book)  # Find every sentence that mentions "Healthcare" exactly as written
findings  # Display the matched sentences
```

```python
# Identical to the previous cell - re-running the same "Healthcare" search again (duplicate cell)
pattern = re. compile("[A-Z]{1}[^.]*[^a-zA-Z]+Healthcare[^a-zA-Z]+[^.]*.")
findings = re.findall(pattern, book)
findings
```

```python
# Case-insensitive version of the "Healthcare" sentence search (catches "healthcare", "HEALTHCARE", etc.)
pattern = re.compile("[^.]*[^a-zA-Z]*Healthcare[^a-zA-Z]+[^.]*.", re.IGNORECASE)
findings = re.findall(pattern, book)  # Find all matching sentences regardless of capitalization
len(findings)  # Count how many sentences mention "Healthcare" in any capitalization
```

# What are the most used words?

```python
# book.split()  # (Commented out) This would split the text into words using whitespace only -
                 # left here as a note/alternative to the regex-based approach used below
```

```python
# Match sequences of one or more letters (a-z or A-Z) - effectively extracting individual "words"
pattern = re.compile("[a-zA-Z]+")
findings = re.findall(pattern, book.lower())  # Lowercase the whole text first, then extract every word
len(findings)  # Count the total number of words found in the entire book
```

```python
# Same word-extraction pattern as above
pattern = re.compile("[a-zA-Z]+")
findings = re.findall(pattern, book.lower())  # Get the list of all lowercase words in the book
findings[:5]  # Look at just the first 5 words as a quick sanity check
```

```python
d = {}  # Create an empty dictionary to hold word -> count pairs
for word in findings:  # Go through every word found in the book, one at a time
    if word in d.keys():  # If we've already seen this word before...
        d[word] = d[word] + 1  # ...increase its count by 1
    else:  # Otherwise, if it's the first time we've seen this word...
        d[word] = 1  # ...add it to the dictionary with a starting count of 1
```

```python
# Turn the dictionary into a list of (count, word) tuples instead of (word, count)
# This makes it possible to sort by count, since Python sorts tuples by their first element first
d_list = [(value, key) for (key, value) in d.items()]
sorted(d_list, reverse=True)  # Sort the list from highest count to lowest, showing the most common words first
```
